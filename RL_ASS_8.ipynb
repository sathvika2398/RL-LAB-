```python

gamma = float(input("Enter discount factor: "))
goal_reward = float(input("Enter goal reward: "))
normal_reward = float(input("Enter normal movement reward: "))
obstacle_reward = float(input("Enter obstacle reward: "))

policy = {
    'S0': 'Right',
    'S1': 'Down',
    'S3': 'Right',
    'S4': 'Up'
}

grid = {
    'S0': (0, 0),
    'S1': (0, 1),
    'G': (0, 2),
    'X': (1, 0),
    'S3': (1, 1),
    'S4': (1, 2)
}

def step(state, action):
    r, c = grid[state]

    if action == 'Up':
        r -= 1
    elif action == 'Down':
        r += 1
    elif action == 'Left':
        c -= 1
    elif action == 'Right':
        c += 1

    if r < 0 or r > 1 or c < 0 or c > 2:
        return state, normal_reward

    next_state = next((s for s, pos in grid.items() if pos == (r, c)), state)

    if next_state == 'X':
        return state, obstacle_reward

    if next_state == 'G':
        return next_state, goal_reward

    return next_state, normal_reward

def state_value_function(state):
    total = 0
    discount = 1
    current = state

    for _ in range(100):
        if current == 'G':
            break
        action = policy[current]
        current, reward = step(current, action)
        total += discount * reward
        discount *= gamma

    return total

def action_value_function(state, action):
    total = 0
    discount = 1
    current = state

    for i in range(100):
        if current == 'G':
            break

        chosen_action = action if i == 0 else policy[current]
        current, reward = step(current, chosen_action)
        total += discount * reward
        discount *= gamma

    return total

state = input("Enter state (S0, S1, S3, S4): ").strip()
action = input("Enter action (Up, Down, Left, Right): ").strip().capitalize()

if state not in policy or action not in ['Up', 'Down', 'Left', 'Right']:
    print("Invalid state or action")
else:
    print("State Value of", state, ":", state_value_function(state))
    print("Action Value of", state, "for", action, ":", action_value_function(state, action))
```
