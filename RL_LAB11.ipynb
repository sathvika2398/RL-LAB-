```python

gamma = float(input("Enter discount factor (gamma): "))
normal_reward = float(input("Enter normal movement reward: "))
wall_reward = float(input("Enter wall reward: "))
goal_reward = float(input("Enter goal reward: "))
theta = float(input("Enter convergence threshold: "))

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", wall_reward),
        "Down": ("S1", normal_reward),
        "Left": ("S0", wall_reward),
        "Right": ("S0", wall_reward)
    },
    "S1": {
        "Up": ("S0", normal_reward),
        "Down": ("S1", wall_reward),
        "Left": ("S1", wall_reward),
        "Right": ("G", goal_reward)
    }
}

V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

policy = {
    "S0": "Up",
    "S1": "Up"
}

policy_iteration = 0

print("\n----- GRID WORLD INFORMATION -----")
print("States:", states + ["G"])
print("Actions:", actions)
print("Initial Policy:", policy)

print("\n----- TRANSITION INFORMATION -----")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

while True:
    policy_iteration += 1

    print("\n----- POLICY ITERATION", policy_iteration, "-----")
    print("Current Policy:", policy)

    evaluation_iteration = 0

    while True:
        new_V = V.copy()

        for state in states:
            action = policy[state]
            next_state, reward = transitions[state][action]
            new_V[state] = reward + gamma * V[next_state]

        delta = max(abs(new_V[state] - V[state]) for state in states)

        V = new_V
        evaluation_iteration += 1

        print("Evaluation", evaluation_iteration,
              "S0 =", round(V["S0"], 6),
              "S1 =", round(V["S1"], 6),
              "Delta =", round(delta, 8))

        if delta < theta:
            break

    new_policy = {}
    Q = {}

    for state in states:
        Q[state] = {}

        for action in actions:
            next_state, reward = transitions[state][action]
            Q[state][action] = reward + gamma * V[next_state]

        new_policy[state] = max(Q[state], key=Q[state].get)

    print("\nImproved Policy:", new_policy)

    if new_policy == policy:
        break

    policy = new_policy

print("\n========== FINAL RESULTS ==========")

print("\nOptimal State Values")

for state in states:
    print("V*(" + state + ") =", round(V[state], 2))

print("V*(G) = 0")

print("\nFinal Action Values")

for state in states:
    print("\nState:", state)

    for action in actions:
        print("Q(" + state + ", " + action + ") =",
              round(Q[state][action], 2))

print("\n========== OPTIMAL POLICY ==========")

for state in states:
    print(state, "->", policy[state])

print("\n========== BELLMAN OPTIMALITY VERIFICATION ==========")

for state in states:
    max_q = max(Q[state].values())

    print("State:", state)
    print("V*(s) =", round(V[state], 2))
    print("Maximum Q value =", round(max_q, 2))
    print("Verification:", abs(V[state] - max_q) < theta)

print("\nOptimal Path: S0 -> S1 -> G")
print("Policy Iterations:", policy_iteration)
print("Policy is stable.")
```
