```python
n = int(input("Enter number of machines: "))

machines = {}

for i in range(n):
    name = input(f"Enter machine {i + 1} name: ")
    rewards = list(map(float, input(f"Enter rewards for {name}: ").split()))
    machines[name] = rewards

highest_machine = None
highest_q = float('-inf')

for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    print(f"\nMachine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q:.2f}")

    if q > highest_q:
        highest_q = q
        highest_machine = name

print(f"\nGreedy Selected Machine: {highest_machine}")
```
