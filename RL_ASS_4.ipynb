```python
import math

machines = ['A', 'B', 'C']
M = 3
N_play = 10

N = [0, 0, 0]
R = [0.0, 0.0, 0.0]
Q = [0.0, 0.0, 0.0]
UCB = [0.0, 0.0, 0.0]

for play in range(1, N_play + 1):
    for i in range(M):
        if N[i] == 0:
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt((2 * math.log(play)) / N[i])

    selected = UCB.index(max(UCB))
    machine = machines[selected]

    print("\nPlay", play)
    print("UCB =", ["inf" if x == float('inf') else round(x, 2) for x in UCB])
    print("Play with machine", machine)

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("N =", N)
    print("R =", R)
    print("Q =", [round(x, 2) for x in Q])

for i in range(M):
    if N[i] == 0:
        UCB[i] = float('inf')
    else:
        UCB[i] = Q[i] + math.sqrt((2 * math.log(N_play)) / N[i])

print("\nFinal Results")
print("-----------------------------")
print("Machine\tN\tR\tUCB\tQ")
print("-----------------------------")

for i in range(M):
    ucb_value = "inf" if UCB[i] == float('inf') else round(UCB[i], 2)
    print(machines[i], "\t", N[i], "\t", R[i], "\t", ucb_value, "\t", round(Q[i], 2))

print("-----------------------------")
print("Total Reward =", sum(R))
```
