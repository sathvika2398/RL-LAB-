```python

gamma = float(input("Enter discount factor (gamma): "))
normal_reward = float(input("Enter normal movement reward: "))
wall_reward = float(input("Enter wall reward: "))
goal_reward = float(input("Enter goal reward: "))
theta = float(input("Enter convergence threshold: "))

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", wall_reward),
        "Down": ("S1", normal_reward),
        "Left": ("S0", wall_reward),
        "Right": ("S0", wall_reward)
    },
    "S1": {
        "Up": ("S0", normal_reward),
        "Down": ("S1", wall_reward),
        "Left": ("S1", wall_reward),
        "Right": ("G", goal_reward)
    }
}

V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

iterations = 0

while True:
    new_V = V.copy()

    for state in states:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            value = reward + gamma * V[next_state]
            values.append(value)

        new_V[state] = max(values)

    delta = max(abs(new_V[state] - V[state]) for state in states)

    V = new_V
    iterations += 1

    if delta < theta:
        break

policy = {}

for state in states:
    action_values = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        action_values[action] = reward + gamma * V[next_state]

    best_action = max(action_values, key=action_values.get)
    policy[state] = best_action

print("\n----- VALUE ITERATION RESULTS -----")

print("Number of iterations:", iterations)

print("\nOptimal State Values:")

for state in states:
    print("V*(" + state + ") =", round(V[state], 2))

print("\nOptimal Action Values:")

for state in states:
    print("\nState:", state)

    for action in actions:
        next_state, reward = transitions[state][action]
        q_value = reward + gamma * V[next_state]
        print("Q(" + state + ", " + action + ") =", round(q_value, 2))

print("\nOptimal Policy:")

for state in states:
    print(state, "->", policy[state])

print("\nConvergence Check:")
print("Final state values:", V)
```
