```python

print("ASSIGNMENT 1: UNDISCOUNTED AND DISCOUNTED RETURNS")

n = int(input("Enter number of time steps: "))
rewards_list = []

for i in range(n):
    r = float(input(f"Enter reward for time step {i + 1}: "))
    rewards_list.append(r)

gamma = float(input("Enter discount factor (gamma): "))

undiscounted_return = sum(rewards_list)
discounted_return = sum((gamma ** i) * rewards_list[i] for i in range(n))

print("Entered Rewards:", rewards_list)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)

print("\nASSIGNMENT 2: 2 x 3 GRID WORLD MDP")

states = ["S0", "S1", "G", "X", "S3", "S4"]
actions = ["Up", "Down", "Left", "Right"]
discount_factor = 0.9

rewards = {
    "S0": -1,
    "S1": -1,
    "G": 10,
    "X": -5,
    "S3": -1,
    "S4": -1
}

transitions = {
    "S0": {"Up": "S0", "Down": "S3", "Left": "S0", "Right": "S1"},
    "S1": {"Up": "S1", "Down": "S1", "Left": "S0", "Right": "G"},
    "G": {"Up": "G", "Down": "G", "Left": "G", "Right": "G"},
    "X": {"Up": "S0", "Down": "X", "Left": "X", "Right": "S3"},
    "S3": {"Up": "S0", "Down": "S3", "Left": "X", "Right": "S4"},
    "S4": {"Up": "G", "Down": "S4", "Left": "S3", "Right": "S4"}
}

print("State Space:", states)
print("Action Space:", actions)
print("Discount Factor:", discount_factor)

state = input("Enter current state (S0, S1, G, X, S3, S4): ")
action = input("Enter action (Up, Down, Left, Right): ")

if state in states and action in actions:
    next_state = transitions[state][action]
    reward = rewards[next_state]

    print("Current State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
    print("Discount Factor:", discount_factor)
else:
    print("Invalid state or action")
```
