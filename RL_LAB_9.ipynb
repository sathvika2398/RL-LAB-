```python

gamma = float(input("Enter discount factor (gamma): "))
r_normal = float(input("Enter normal movement reward: "))
r_wall = float(input("Enter wall reward: "))
r_goal = float(input("Enter goal reward: "))

p_s0_down = float(input("Enter probability of S0 Down: "))
p_s0_right = float(input("Enter probability of S0 Right: "))
p_s1_right = float(input("Enter probability of S1 Right: "))
p_s1_up = float(input("Enter probability of S1 Up: "))

q_s0_right = r_wall
q_s1_right = r_goal

a = p_s0_down * gamma
b = p_s1_up * gamma
c = p_s1_right * gamma

constant1 = p_s0_down * r_normal + p_s0_right * q_s0_right
constant2 = p_s1_right * r_goal + p_s1_up * r_normal

v_s0 = (constant1 + a * constant2) / (1 - a * b)
v_s1 = constant2 + b * v_s0

q_s0_down = r_normal + gamma * v_s1
q_s0_right = r_wall
q_s1_right = r_goal
q_s1_up = r_normal + gamma * v_s0

print("\n----- ASSIGNMENT 1 RESULTS -----")
print("V(S0) =", round(v_s0, 2))
print("V(S1) =", round(v_s1, 2))
print("Q(S0, Down) =", round(q_s0_down, 2))
print("Q(S0, Right) =", round(q_s0_right, 2))
print("Q(S1, Right) =", round(q_s1_right, 2))
print("Q(S1, Up) =", round(q_s1_up, 2))

print("\nVerification")
print("V(S0) =", round(p_s0_down * q_s0_down + p_s0_right * q_s0_right, 2))
print("V(S1) =", round(p_s1_right * q_s1_right + p_s1_up * q_s1_up, 2))
```

```python

gamma = float(input("Enter discount factor (gamma): "))
theta = float(input("Enter convergence threshold: "))

r_normal = float(input("Enter normal movement reward: "))
r_wall = float(input("Enter wall reward: "))
r_goal = float(input("Enter goal reward: "))

p_s0_wall = float(input("Enter probability of S0 Down to Wall: "))
p_s0_s1 = float(input("Enter probability of S0 Down to S1: "))

p_s1_goal = float(input("Enter probability of S1 Right to Goal: "))
p_s1_s0 = float(input("Enter probability of S1 Right to S0: "))

v_s0 = 0.0
v_s1 = 0.0
iterations = 0

while True:
    new_v_s0 = p_s0_wall * (r_wall + gamma * 0) + p_s0_s1 * (r_normal + gamma * v_s1)

    new_v_s1 = p_s1_goal * (r_goal + gamma * 0) + p_s1_s0 * (r_normal + gamma * v_s0)

    delta = max(abs(new_v_s0 - v_s0), abs(new_v_s1 - v_s1))

    v_s0 = new_v_s0
    v_s1 = new_v_s1

    iterations += 1

    if delta < theta:
        break

q_s0_down = p_s0_wall * (r_wall + gamma * 0) + p_s0_s1 * (r_normal + gamma * v_s1)

q_s0_right = r_wall + gamma * 0

q_s1_right = p_s1_goal * (r_goal + gamma * 0) + p_s1_s0 * (r_normal + gamma * v_s0)

q_s1_up = r_normal + gamma * v_s0

print("\n----- ASSIGNMENT 2 RESULTS -----")
print("Number of iterations:", iterations)
print("V(S0) =", round(v_s0, 2))
print("V(S1) =", round(v_s1, 2))

print("\nAction Values")
print("Q(S0, Down) =", round(q_s0_down, 2))
print("Q(S0, Right) =", round(q_s0_right, 2))
print("Q(S1, Right) =", round(q_s1_right, 2))
print("Q(S1, Up) =", round(q_s1_up, 2))

print("\nPolicy Consistency Check")
print("V(S0) =", round(v_s0, 6))
print("Q(S0, Down) =", round(q_s0_down, 6))
print("V(S1) =", round(v_s1, 6))
print("Q(S1, Right) =", round(q_s1_right, 6))

print("\nVerification:")
print("S0:", abs(v_s0 - q_s0_down) < theta)
print("S1:", abs(v_s1 - q_s1_right) < theta)
```
