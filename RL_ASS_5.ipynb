```python
import gymnasium as gym

print("Available Environments")
print("1. CartPole")
print("2. MountainCar")
print("3. FrozenLake")

choice = int(input("Enter environment choice (1-3): "))
episodes = int(input("Enter number of episodes: "))
max_steps = int(input("Enter maximum steps per episode: "))

if choice == 1:
    env = gym.make("CartPole-v1")
    environment_name = "CartPole"
elif choice == 2:
    env = gym.make("MountainCar-v0")
    environment_name = "MountainCar"
elif choice == 3:
    env = gym.make("FrozenLake-v1", is_slippery=False)
    environment_name = "FrozenLake"
else:
    print("Invalid choice")
    exit()

total_reward_all = 0

print("\nSelected Environment:", environment_name)

for episode in range(1, episodes + 1):
    state, info = env.reset()
    total_reward = 0

    print("\nEpisode", episode)
    print("Initial State:", state)

    for step in range(1, max_steps + 1):
        print("Step", step)
        print("Available Actions:", env.action_space)

        action = env.action_space.sample()
        print("Selected Action:", action)

        next_state, reward, terminated, truncated, info = env.step(action)

        print("Next State:", next_state)
        print("Reward:", reward)

        total_reward += reward

        if terminated or truncated:
            print("Episode terminated")
            break

        state = next_state

    print("Episode Total Reward:", total_reward)
    total_reward_all += total_reward

average_reward = total_reward_all / episodes

print("\nFinal Results")
print("----------------------------")
print("Environment:", environment_name)
print("Number of Episodes:", episodes)
print("Total Reward:", total_reward_all)
print("Average Reward:", average_reward)
print("----------------------------")

env.close()
```
