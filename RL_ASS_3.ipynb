```python
import random

machines = ['A', 'B', 'C']
M = 3
N_play = 10
epsilon = 0.2

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

random.seed(10)

for play in range(1, N_play + 1):
    print("\nPlay", play)

    r = random.random()
    best = Q.index(max(Q))

    if r < epsilon:
        choices = [i for i in range(M) if i != best]
        selected = random.choice(choices)
        print("Exploration")
    else:
        selected = best
        print("Exploitation")

    machine = machines[selected]

    print("Play with machine", machine)
    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("N =", N)
    print("R =", R)
    print("Q =", [round(x, 2) for x in Q])

print("\nFinal Results")
print("-------------------------")
print("Machine\tN\tR\tQ")
print("-------------------------")

for i in range(M):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", round(Q[i], 2))

print("-------------------------")
print("Total Reward =", sum(R))
```
