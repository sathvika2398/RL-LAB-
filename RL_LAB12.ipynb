```python

gamma = float(input("Enter discount factor (gamma): "))
normal_reward = float(input("Enter normal movement reward: "))
wall_reward = float(input("Enter wall reward: "))
goal_reward = float(input("Enter goal reward: "))
theta = float(input("Enter convergence threshold: "))

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", wall_reward),
        "Down": ("S1", normal_reward),
        "Left": ("S0", wall_reward),
        "Right": ("S0", wall_reward)
    },
    "S1": {
        "Up": ("S0", normal_reward),
        "Down": ("S1", wall_reward),
        "Left": ("S1", wall_reward),
        "Right": ("G", goal_reward)
    }
}

V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

iteration = 0

print("\n----- GRID WORLD INFORMATION -----")
print("States:", states + ["G"])
print("Actions:", actions)
print("Initial Values:", V)

print("\n----- TRANSITION INFORMATION -----")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

print("\n----- VALUE ITERATION PROCESS -----")
print("Iteration\tV(S0)\t\tV(S1)\t\tDelta")

while True:
    new_V = V.copy()

    for state in states:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[state] - V[state]) for state in states)

    V = new_V
    iteration += 1

    print(iteration, "\t\t", round(V["S0"], 6),
          "\t", round(V["S1"], 6),
          "\t", round(delta, 8))

    if delta < theta:
        break

Q = {}
policy = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        Q[state][action] = reward + gamma * V[next_state]

    policy[state] = max(Q[state], key=Q[state].get)

print("\n----- FINAL OPTIMAL STATE VALUES -----")

for state in states:
    print("V*(" + state + ") =", round(V[state], 2))

print("V*(G) = 0")

print("\n----- FINAL ACTION VALUES -----")

for state in states:
    print("\nState:", state)

    for action in actions:
        next_state, reward = transitions[state][action]
        print("Q(" + state + ", " + action + ") =",
              round(Q[state][action], 2))

print("\n----- OPTIMAL POLICY -----")

for state in states:
    print(state, "->", policy[state])

print("\n----- BELLMAN OPTIMALITY VERIFICATION -----")

for state in states:
    max_q = max(Q[state].values())

    print("State:", state)
    print("V*(s) =", round(V[state], 2))
    print("Maximum Q(s,a) =", round(max_q, 2))
    print("Verified:", abs(V[state] - max_q) < theta)

print("\nNumber of Iterations:", iteration)
print("Optimal Path: S0 -> S1 -> G")
```
