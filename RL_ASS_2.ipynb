```python
import random
machines = ['A', 'B', 'C']
M = 3
N_play = 10
epsilon = 0.2
N = [0, 0, 0]
R = [[], [], []]
Q = [0.0, 0.0, 0.0]
for play in range(1, N_play + 1):
    r = random.random()
    if r < epsilon:
        choice = random.randint(0, M - 1)
        mode = "Exploration"
    else:
        max_q = max(Q)
        best = [i for i in range(M) if Q[i] == max_q]
        choice = random.choice(best)
        mode = "Exploitation"
    print("\nPlay", play)
    print("Mode:", mode)
    print("Selected machine:", machines[choice])
    reward = float(input("Enter reward: "))
    N[choice] += 1
    R[choice].append(reward)
    Q[choice] = Q[choice] + (reward - Q[choice]) / N[choice]
    print("Updated Q-value:", Q[choice])
total_reward = sum(sum(rewards) for rewards in R)
print("\nFinal Results")
print("Machine\tN\tRewards\t\tQ-value")
for i in range(M):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", round(Q[i], 2))
print("\nTotal Reward:", total_reward)
```
