```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
n_play = 10

N = [0, 0, 0]
R = [[], [], []]
Q = [0.0, 0.0, 0.0]

total_reward = 0

for play in range(1, n_play + 1):
    r = random.random()

    if r < epsilon:
        selected = random.choice(range(3))
        decision = "Exploration"
    else:
        selected = Q.index(max(Q))
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.2f}")
    print(f"Decision = {decision}")
    print(f"Selected Machine = {machines[selected]}")

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected].append(reward)
    total_reward += reward

    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print(f"N({machines[selected]}) = {N[selected]}")
    print(f"Q({machines[selected]}) = {Q[selected]:.2f}")

print("\nFinal Results")
print("Machine\tN\tRewards\t\tQ")

for i in range(3):
    print(f"{machines[i]}\t{N[i]}\t{R[i]}\t{Q[i]:.2f}")

print(f"\nTotal Reward = {total_reward:.2f}")

selected = Q.index(max(Q))
print(f"Best Machine = {machines[selected]}")
print(f"Highest Q-value = {Q[selected]:.2f}")
```
