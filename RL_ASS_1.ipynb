```python
n = int(input("Enter number of machines: "))
machines = []
q_values = []
for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(float, input("Enter rewards separated by space: ").split()))
    q = sum(rewards) / len(rewards)
    machines.append((name, rewards))
    q_values.append(q)
for i in range(n):
    name, rewards = machines[i]
    print("Machine", name)
    print("Rewards:", [int(x) if x.is_integer() else x for x in rewards])
    print(f"Q({name}) = {q_values[i]:.2f}")
selected = machines[q_values.index(max(q_values))][0]
print("Greedy Selected Machine:", selected)
```

```python
n = int(input("Enter number of machines: "))
machines = []
rewards = {}
for i in range(n):
    name = input("Enter machine name: ")
    r = list(map(int, input("Enter rewards separated by space: ").split()))
    machines.append(name)
    rewards[name] = r
q = {name: 0.0 for name in machines}
count = {name: 0 for name in machines}
for name in machines:
    for r in rewards[name]:
        count[name] += 1
        q[name] = q[name] + (r - q[name]) / count[name]
selected = max(q, key=q.get)
print("\nFinal Q Values:")
for name in machines:
    print(f"Q({name}) = {q[name]:.2f}")
print("Greedy Selected Machine:", selected)
```
