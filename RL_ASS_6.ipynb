```python

!pip install -q -U gymnasium

import gymnasium as gym

print("1. CliffWalking")
print("2. Taxi")

choice = int(input("Enter environment choice (1 or 2): "))
episodes = int(input("Enter number of episodes: "))

if choice == 1:
    env = gym.make("CliffWalking-v1")
elif choice == 2:
    env = gym.make("Taxi-v3")
else:
    raise ValueError("Invalid choice. Enter 1 or 2.")

all_rewards = []

for episode in range(1, episodes + 1):
    state, info = env.reset()
    total_reward = 0
    steps = 0
    terminated = False
    truncated = False

    while not (terminated or truncated):
        action = env.action_space.sample()
        next_state, reward, terminated, truncated, info = env.step(action)

        total_reward += reward
        steps += 1

        print("Episode:", episode,
              "State:", state,
              "Action:", action,
              "Reward:", reward,
              "Next State:", next_state)

        state = next_state

    average_reward = total_reward / steps if steps > 0 else 0
    all_rewards.append(total_reward)

    print("\nEpisode:", episode)
    print("Total Reward:", total_reward)
    print("Average Reward:", average_reward)
    print("Total Steps:", steps)
    print("-" * 40)

print("\nAverage Reward Across All Episodes:",
      sum(all_rewards) / len(all_rewards))

env.close()
```
